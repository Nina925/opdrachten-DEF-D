#### Voeg hier een foto van de schets van je plaatcondensator in. Geef daarbij ook duidelijk de waarde van $A$ en $d$ aan.

```python
# Schrijf hier een python functie die de capaciteit van een plaatcondensator uitrekent. Zorg dat de functie
# als input de ontwerpparameters van de vlakke plaat formule accepteert (oppervlakte, afstand en diëlektrische
# constante) en als output de capaciteit van de vlakke plaat condensator met die ontwerpparameters.
# Je gaat deze formule de rest van de dag gebruiken: controleer hem goed!
import numpy as np

epsilon0 = 8.854e-12  # 

def berekening_condensator(epsilonr,A,d):
    epsilonr = np.array(epsilonr)
    A = np.array(A)
    d = np.array(d)
    
    return (epsilon0 * epsilonr * A) / d
    
epsilonr_values = [3]
A_values = [0.01]  
d_values = [0.001]
    
    
# Optioneel: schrijf je functie zodanig dat je arrays als input kan geven en (dus) ook arrays als output.
```

```python
# Gebruik je formule om uit te rekenen wat je verwacht dat de capaciteit van jullie vlakke plaat condensatoren
# gaat zijn.

C = berekening_condensator(epsilonr_values, A_values, d_values)
C_pF = C * 1e12

print(f"Berekende capaciteit plaatcondensator groep 14 is {C_pF[0]:.1f} pF")

```

## Opdracht 7: Plot de kalibratiegrafiek.

```python
# Plot hier je gemeten kalibratiegrafiek en plot in dezelfde figuur ook je verwachte ontwerpgrafiek.
# Gemeten Arduino data (omgezet naar kg)
gewicht1 = np.array([539.2,999.2,451.2,911.2,378.2,838.2,271.7,731.7]) * 1e-3  # kg
capaciteit1 = np.array([10.6,11.7,10.0,11.0,9.9,10.8,9.7,10.6])                 # pF

# Sorteer gemeten data op gewicht
sort_index = np.argsort(gewicht1)
gewicht1_sorted = gewicht1[sort_index]
capaciteit1_sorted = capaciteit1[sort_index]           


# Plot
plt.figure(figsize=(7,5))
plt.plot(gewicht,C2, marker = 'o', linestyle='-')
plt.plot(gewicht1_sorted, capaciteit1_sorted, marker='s', linestyle='--', label='Gemeten capaciteit')

# Labels, titel en legend
plt.xlabel('Gewicht (kg)')
plt.ylabel('Capaciteit (pF)')
plt.title('Kalibratiegrafiek: Gewicht versus Capaciteit')
plt.legend()
plt.grid(True)
```
